```python
import os
import pandas as pd

def bulk_rename_files(directory_path, csv_path):
    # Load the CSV file
    try:
        df = pd.read_csv(csv_path)
    except Exception as e:
        print(f"Error reading CSV: {e}")
        return

    # Create a mapping dictionary for quick lookup {id: new_title}
    # Using strings to ensure matching works correctly even if IDs look like numbers
    rename_map = dict(zip(df['id'].astype(str), df['new_title'].astype(str)))

    # Track how many files were renamed
    renamed_count = 0
    skipped_count = 0

    # Iterate through files in the specified directory
    for filename in os.listdir(directory_path):
        # Split extension and base filename
        base_name, extension = os.path.splitext(filename)
        
        # We only care about .tif and .gcps as per your request
        if extension.lower() in ['.tif', '.csv']:
            # Check if the base_name exists in our mapping
            if base_name in rename_map:
                new_base_name = rename_map[base_name]
                new_filename = f"{new_base_name}{extension}"
                
                old_path = os.path.join(directory_path, filename)
                new_path = os.path.join(directory_path, new_filename)
                
                # Perform the rename
                try:
                    os.rename(old_path, new_path)
                    print(f"Renamed: {filename} -> {new_filename}")
                    renamed_count += 1
                except Exception as e:
                    print(f"Failed to rename {filename}: {e}")
            else:
                skipped_count += 1

    print(f"\nProcessing complete.")
    print(f"Files renamed: {renamed_count}")
    print(f"Files skipped (no match in CSV): {skipped_count}")

if __name__ == "__main__":
    # Path to the directory containing your .tif and .gcps files
    target_directory = "E:/Users/brodeujj/mapwarper" 
    
    # Path to your CSV file
    csv_file_path = "E:/Users/brodeujj/mapwarper/all_data-rename.csv"
    
    bulk_rename_files(target_directory, csv_file_path)
```
